Analyzing the Data

```python
print(np.mean(data))
```

```python
#not all functions have input that's why
import time
print(time.ctime())
```

```python
maxval, minval, stdval = np.amax(data), np.amin(data), np.std(data)

print('maximum inflammation:', maxval)
print('minimum inflammation:', minval)
print('standard deviation:', stdval)
```

```python
#0 will be on the first axis (rows), while everything on the second (columns)
patient_0 = data[0, :]
print('maximum inflammation for patient 0:', np.amax(patient_0))
```

```python
print('maximum inflammation for patient 2:', np.amax(data[2, :]))
```

```python
print(np.max(data, axis=1))
```

```python
print(np.max(data, axis=1).shape)
```

```python
print(np.mean(data, axis=0))
```

```python
print(np.mean(data, axis=0).shape)
```

```python
print(np.mean(data, axis=1))
```

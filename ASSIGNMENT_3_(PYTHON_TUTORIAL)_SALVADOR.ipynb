```python
#uploading the fits file data
img_name = 'pdi_pi_collapsed.fits'
img = fits.getdata(img_name, ext=1)
```

```python
#visualization of the data (plotting)
fig = plt.figure(figsize=(10,10))
image = plt.imshow(img, origin = 'lower', vmin=-0.001, vmax=0.01)
cbar = fig.colorbar(image,shrink=0.9)
cbar.set_label('Surface Brightness (arb. units)', rotation=90,fontsize=12)
plt.xlim([60,140])
plt.ylim([60,140])
plt.xlabel(r'$\Delta$ RA [px]',fontsize=12)
plt.ylabel(r'$\Delta$ Dec [px]',fontsize=12);
```

```python
#uploading the csv data file
stellar=pd.read_csv('closest20kstars.csv')

#creating plt
fig = plt.figure(figsize = [10,10])

#will create the scatter plot
plt.scatter(stellar['ra'], stellar['dec'], alpha=0.8, s=12, lw = 0.5, ec = 'g', color='pink')
#labeling the elements
plt.xlim([0,362])
plt.ylim([-75,75])
plt.xlabel('RA [$\degree$]', fontsize = 14)
plt.ylabel('Dec [$\degree$]', fontsize = 14)
plt.title('Closest Stars 20,000 stars from Gaia Archive (DR2)', fontsize = 16)
```

Fibonacci Sequence

```python
def fibonacci_sequence(n):
  "Prints the Fibonacci sequence up to n terms."
  a, b = 0, 1
  for _ in range(n):
    print(a)
    a, b = b, a + b

fibonacci_sequence(20) #print the first 10 of the Fibonacci sequence
```

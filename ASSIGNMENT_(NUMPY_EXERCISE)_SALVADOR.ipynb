```python
x = np.linspace(0, 10, 100)
y = np.random.rand(100)
z = x+y
print(z)
```

```python
sum_x = x.sum()
sum_y = y.sum()
sum_z = z.sum()
print("The Sum of x is", sum_x)
print("The Sum of y is", sum_y)
print("The Sum of z is", sum_z)
largest = max(sum_x, sum_y, sum_z)
print("The Largest sum is", largest)
```
